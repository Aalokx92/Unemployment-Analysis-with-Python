```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings

warnings.filterwarnings("ignore")
```

```python
file_path = "/Users/aalok_x92/Downloads/Unemployment in India.csv"

df = pd.read_csv(file_path)
```

```python
print("\n========== FIRST 5 ROWS ==========\n")
print(df.head())
```

```python
print("\n========== COLUMN NAMES ==========\n")
print(df.columns.tolist())
```

```python
df.columns = df.columns.str.strip()
```

```python
print("\n========== CLEANED COLUMN NAMES ==========\n")
print(df.columns.tolist())
```

```python
print("\n========== DATAFRAME ==========\n")
print(df)
```

```python
print("\n========== DATASET SIZE ==========\n")

print(
    f"The dataframe has {df.shape[0]} rows "
    f"and {df.shape[1]} columns"
)
```

```python
print("\n========== DATA INFORMATION ==========\n")

df.info()
```

```python
print("\n========== STATISTICAL SUMMARY ==========\n")

print(df.describe())
```

```python
print("\n========== MISSING VALUES ==========\n")

print(df.isnull().sum())
```

```python
df = df.dropna()


print("\n========== MISSING VALUES AFTER CLEANING ==========\n")

print(df.isnull().sum())
```

```python
print("\n========== DUPLICATE ROWS ==========\n")

print(
    "Number of duplicate rows:",
    df.duplicated().sum()
)
```

```python
df = df.drop_duplicates()


print(
    "Shape after removing duplicates:",
    df.shape
)
```

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    dayfirst=True
)
```

```python
df["Day"] = df["Date"].dt.day
```

```python
df["Month"] = df["Date"].dt.month_name()
```

```python
df["Year"] = df["Date"].dt.year
```

```python
print("\n========== DATA AFTER DATE PROCESSING ==========\n")

print(df.head())
```

```python
print("\n========== DESCRIPTIVE STATISTICS ==========\n")

print(df.describe())
```

```python
mean_unemployment = (
    df["Estimated Unemployment Rate (%)"].mean()
)


print(
    f"\nMean Unemployment Rate: "
    f"{mean_unemployment:.2f}%"
)
```

```python
median_unemployment = (
    df["Estimated Unemployment Rate (%)"].median()
)


print(
    f"Median Unemployment Rate: "
    f"{median_unemployment:.2f}%"
)
```

```python
std_unemployment = (
    df["Estimated Unemployment Rate (%)"].std()
)


print(
    f"Standard Deviation of Unemployment Rate: "
    f"{std_unemployment:.2f}%"
)
```

```python
mean_unemployment_over_time = (
    df.groupby("Date")["Estimated Unemployment Rate (%)"]
    .mean()
)


plt.figure(figsize=(10, 6))


plt.plot(
    mean_unemployment_over_time.index,
    mean_unemployment_over_time.values,
    marker="o",
    linestyle="-"
)


plt.xlabel("Date")

plt.ylabel("Estimated Unemployment Rate (%)")

plt.title("Unemployment Rate Over Time")

plt.grid(True)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
region_unemployment = (
    df.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)


fig = px.bar(
    region_unemployment,
    x="Region",
    y="Estimated Unemployment Rate (%)",
    color="Region",
    title="Average Unemployment Rate by Region"
)


fig.update_layout(
    xaxis={
        "categoryorder": "total descending"
    }
)


fig.show()
```

```python
monthly_employment = (
    df.groupby("Month")["Estimated Employed"]
    .mean()
    .reset_index()
)
```

```python
month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]


monthly_employment["Month"] = pd.Categorical(
    monthly_employment["Month"],
    categories=month_order,
    ordered=True
)


monthly_employment = monthly_employment.sort_values(
    "Month"
)


fig = px.bar(
    monthly_employment,
    x="Month",
    y="Estimated Employed",
    color="Month",
    title="Average Estimated Employed by Month"
)

fig.show()
```

```python
sunburst_data = (
    df.groupby(
        ["Area", "Region"]
    )["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)


fig = px.sunburst(
    sunburst_data,
    path=["Area", "Region"],
    values="Estimated Unemployment Rate (%)",
    title="Unemployment Rate by Area and Region"
)


fig.show()
```

```python
plt.figure(figsize=(14, 7))


sns.boxplot(
    x="Region",
    y="Estimated Unemployment Rate (%)",
    data=df
)


plt.xticks(rotation=90)

plt.title(
    "Box Plot of Estimated Unemployment Rate by Region"
)

plt.xlabel("Region")

plt.ylabel("Estimated Unemployment Rate (%)")

plt.tight_layout()

plt.show()
```

```python
correlation_columns = [
    "Estimated Unemployment Rate (%)",
    "Estimated Employed",
    "Estimated Labour Participation Rate (%)"
]


correlation = df[correlation_columns].corr()


print("\n========== CORRELATION MATRIX ==========\n")

print(correlation)
```

```python
plt.figure(figsize=(8, 6))


sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)


plt.title("Correlation Heatmap")

plt.tight_layout()

plt.show()
```

```python
from scipy.stats import ttest_ind


urban_unemployment = df[
    df["Area"] == "Urban"
]["Estimated Unemployment Rate (%)"]


rural_unemployment = df[
    df["Area"] == "Rural"
]["Estimated Unemployment Rate (%)"]


t_stat, p_val = ttest_ind(
    urban_unemployment,
    rural_unemployment,
    equal_var=False
)


print("\n========== URBAN VS RURAL T-TEST ==========\n")


print(
    f"T-Statistic: {t_stat:.2f}"
)


print(
    f"P-Value: {p_val:.4f}"
)


if p_val < 0.05:

    print(
        "Result: There is a statistically significant "
        "difference between Urban and Rural unemployment."
    )

else:

    print(
        "Result: There is no statistically significant "
        "difference between Urban and Rural unemployment."
    )
```
